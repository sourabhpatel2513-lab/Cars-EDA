## Feature Engineering (transforming the columns as per the requirement)

###### Spliiting the Mileage column in to Mileage_value and Mileage_unit

```python
df[['Mileage_value','Mileage_unit']]= df['Mileage'].str.split(' ',expand=True)
```

###### Spliiting the Engine column in to Engine_value and Engine_unit


```python
df[['Engine_value','Engine_unit']]= df['Engine'].str.split(' ',expand=True)
```

###### Spliiting the Power column in to Power_value and Power_unit

```python
df[['Power_value','Power_unit']]= df['Power'].str.split(' ',expand=True)
```

```python
df.head()
```

##### Drop the Mileage,Engine and Power column

```python
df.columns
```

```python
df.drop(columns=['Mileage', 'Engine', 'Power'],inplace=True)
```

```python
df.columns
```

```python
df.head()
```

###### Checking the data types

```python
df.dtypes
```

###### Changing the data type of the Mileage_value,Engine_value,Power_value in numeric data

```python
df['Mileage_value']= df['Mileage_value'].astype('float')
df['Mileage_value'].dtype
```

###### .astype() doesn't work with the column having missing values

```python
df['Engine_value'] = pd.to_numeric(df['Engine_value'])
df['Engine_value'].dtype
```

```python
df['Power_value'] =  pd.to_numeric(df['Power_value'],errors='coerce')
```

##### Check the null values

```python
df.isnull().sum()
```

```python
df.columns
```

```python
num = ['Kilometers_Driven', 'Mileage_value', 'Engine_value','Power_value']
filled = ['Company_name', 'Model_name', 'Fuel_Type', 'Price']
```

```python
cat = []
for i in df.columns:
    if i not in num and i not in filled:
        cat.append(i)
cat
```

##### Filling the null values
By mode: for the categorical data

```python
cat
```

```python
for i in cat:
    df[i].fillna(df[i].mode()[0], inplace = True)
```

```python
for i in cat:
    print(i, ":", df[i].isnull().sum())
```

##### Count the no of cars by year

```python
x = df['Year'].value_counts().index
y = df['Year'].value_counts().values
bars = plt.bar(x,y)
plt.bar_label(bars, label = y)
plt.xticks(rotation = 90)
plt.show()
```
